# Feature engineering

Create features from information available at the current observation and label each row with the next observed active-power value. The horizon is one row, not always one minute; the local sample contains an observed two-minute timestamp gap.

```python
from pathlib import Path
import sys

source_dirs = [Path.cwd() / 'src', Path.cwd().parent / 'src']
source_dir = next(path for path in source_dirs if (path / 'ml' / 'data_loader.py').is_file())
sys.path.insert(0, str(source_dir))
from ml.data_loader import load_processed_data
from ml.features import FEATURE_COLUMNS, make_next_observation_dataset

data = load_processed_data()
timestamp_intervals = data['timestamp'].diff().dropna()
print('Observed intervals:\n', timestamp_intervals.value_counts().to_string())
samples = make_next_observation_dataset(data)
print('Model features:', FEATURE_COLUMNS)
print('Training examples:', len(samples))
display(samples.head())
```

`previous_active_power` is the active-power reading at the feature timestamp; the target is the next row's reading. Calendar features also refer to the feature timestamp. Other measurements from the current row are excluded, and no future row is used as an input. The final row has no subsequent label and is omitted by the one-step shift. Verify timestamp intervals before describing the row-ahead task as minute-ahead forecasting.
